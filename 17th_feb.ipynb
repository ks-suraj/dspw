```python
# Q1. What is MongoDB? Explain non-relational databases in short. In which scenarios it is preferred to use MongoDB over SQL databases?
# Ans : MongoDB : MongoDB is a NoSQL database that stores data in flexible, JSON-like documents. Unlike SQL databases, it doesn't require a fixed schema, allowing for easier handling of unstructured data.
#     : NoSQL : NoSQL databases, like MongoDB, don't use tables and schemas. Instead, they store data in formats like documents, key-value pairs, or graphs, offering flexibility and scalability.
#     : When to Use MongoDB Over SQL
#       Flexible Schema: Ideal for changing or unstructured data.
#       Big Data: Handles large-scale data and real-time processing.
#       Scalability: Supports horizontal scaling across servers.
#       High Traffic: Designed for high read/write workloads.
```

```python
# Q2. State and Explain the features of MongoDB.
# Ans : MongoDB offers several key features:

# Flexible Document Model: Stores data in flexible BSON documents, allowing varying structures.
# Schema-less: No fixed schema is required, making it ideal for evolving data models.
# Horizontal Scalability: Supports sharding to distribute data across multiple servers for scalability.
# Indexing: Allows indexing on any field to enhance query performance.
# High Availability: Uses replication to ensure data redundancy and fault tolerance.
# Aggregation Framework: Provides powerful data processing tools for filtering and grouping.
# Ad-Hoc Queries: Supports dynamic and complex queries.
# Load Balancing: Distributes load across servers to prevent bottlenecks.
# File Storage (GridFS): Efficiently stores large files by splitting them into chunks.
# Strong Consistency: Ensures data is immediately readable after being written.
```
