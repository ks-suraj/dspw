```python
# Q1. Which function is used to open a file? What are the different modes of opening a file? Explain each mode of file opening.
# Ans : open() function is used to open the file
#     : eg : <variable_name>=open.("<file_name>",'<mode>')
#     : modes : 
#       'r' (read mode) : This is the default mode for opening a file. It opens the file for reading and the file pointer is placed at the beginning of the file.
#       'w' (write mode) : This mode opens the file for writing and the file pointer is placed at the beginning of the file. If the file exists, it is truncated (emptied). If the file does not exist, it is created.
#       'a' (append mode) : This mode opens the file for writing and the file pointer is placed at the end of the file. If the file exists, new data is appended to the end of the file. If the file does not exist, it is created.
#       'r+' (read and write mode) : This mode opens the file for both reading and writing and the file pointer is placed at the beginning of the file.
#       'w+' (write and read mode) : This mode opens the file for both writing and reading and the file pointer is placed at the beginning of the file. If the file exists, it is truncated (emptied). If the file does not exist, it is created.
#       'a+' (append and read mode) : This mode opens the file for both writing and reading and the file pointer is placed at the end of the file. If the file exists, new data is appended to the end of the file. If the file does not exist, it is created.
#       'b' (binary mode) : This mode opens the file in binary mode. This is useful for working with binary files, such as images and audio files.

# eg :
f=open("text1.txt",'w') # write mode
f.write("this is a text file\n")
f.close()
f1=open("text1.txt",'r') # read mode
print(f1.read())
f1.close()
f2=open("text1.txt",'w+') # read and write mode
f2.write("an another line\n")
f2.seek(0)
print(f2.read())
f2.close()
f3=open("text1.txt",'r+') #write and read mode
f3.write("an another line using another mode\n")
f3.seek(0)
print(f3.read())
f3.close()
f4=open("text1.txt",'a') # append mode
f4.write("an another line using append\n")
f4.close()
f5=open("text1.txt",'a+') # append and read mode
f5.write("an another line using a+")
f5.seek(0)
print(f5.read())
f5.close()
fg=open("tet2.bin",'wb+') # binary mode
fg.write(b"825")
fg.seek(0)
print(fg.read())
fg.close()
```

```python
# Q3. Write a python program to create a text file. Write ‘I want to become a Data Scientist’ in that file. Then close the file. Open this file and read the content of the file.
with open("txt1.txt",'w+') as f :
    f.write("i want to become a data scientist.")
    f.close()
with open("txt1.txt",'r') as f1 :
    print(f1.read())
    f1.close()
```

```python
# Q4. Explain the following with python code: read(), readline() and readlines().
# Ans : read() : read the entire content of the file from the cursor begins.
#     : readline() : reads a single line from the file including the newline charecter.
#     : readlines() : reads all the lines from the files.
# eg :
with open("text1.txt",'r') as f :
    print(f.read())
    f.seek(0)
    print(f.readline())
    print(f.readlines())
    f.close()
```
