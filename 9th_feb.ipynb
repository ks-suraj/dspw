```python
# Q2. Create a child class car from the vehicle class created in Que 1, which will inherit the vehicle class. Create a method named seating_capacity which takes capacity as an argument and returns the name of the vehicle and its seating capacity.
class vehicle :
    def __init__(self,name_of_vehicle,max_speed,average_of_vehicle):
        self.name_of_vehicle=name_of_vehicle
        self.max_speed=max_speed
        self.average_of_vehicle=average_of_vehicle
a=vehicle("BMW","350mph","280mph")
class car(vehicle):
    def seating_capacity(self,x):
        self.x=x
        print("name of vehicle :",self.name_of_vehicle)
        print("seating capacity :",self.x)
b=car("merc","250mph","200mph")
i=int(input("enter the seating capacity"))
b.seating_capacity(i)
```

```python
# Q3. What is multiple inheritance? Write a python code to demonstrate multiple inheritance.
# Ans : Multiple Inheritance is a type of inheritance which allows a class to inherit the properties from multiple classes in a class.
# eg :
class punch():
    def can_punch(self):
        print("ability to punch")
class kick():
    def can_kick(self):
        print("ability to kick")
class fight(punch,kick): # fight class has inherited from punch and fight classes, hence multiple inheritance
    def can_block(self):
        print("ability to block")
        self.can_punch()
        self.can_kick()
c=fight()
c.can_block()
```

```python
# Q4. What are getter and setter in python? Create a class and create a getter and a setter method in this class.
# Ans : Getter method is a way of seeking or accessing the attributes of a class.
#     : Setter method is a way of modifying or assigning or mutating the attribute values.
class trial:
    def __init__(self,attr=0):
        self.attr=attr
    def get_attr(self): # getter method
        return self.attr
    def set_attr(self,x): # setter method
        self.attr=x
a=trial()
print(a.get_attr())
a.set_attr(3)
print(a.get_attr())
```

```python
# Q5.What is method overriding in python? Write a python code to demonstrate method overriding.
# Ans : Method Overriding : Method Overriding is a concept where a method is inherited from the parent but its definition in the child class is rewritten and is quite different from its parent method.
class example1 :
    def __init__(self,item):
        self.item=item
    def display(self,i):
        self.i=i
        print("inside parent",self.i)
class example2 (example1):
    def __init__(self,var):
        self.var=var
    def display(self,j):
        self.j=j
        print("inside child",self.j)
b=example1(3)
b.display(5)
a=example2(2)
a.display(4)
```
