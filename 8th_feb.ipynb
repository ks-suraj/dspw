```python
# Q3. What is abc module in python? Why is it used?
# Ans : ABC stands for abstract base classes which are a special type of classes, they cannot be instantiated but inherited, they help provide a platform where set of classes can be defined and also they provide with an advantage where all the methods need not be implemented in base class.
#     : abstract class helps create a skeleton or bluprint of a model or code
#     : abstract class also helps in modifying the methods in seperate classes which are being inherited from the base class.
from abc import ABC
class base(ABC):
    @abc.abstractmethod
    def define(self):
        pass
class string(base):
    def define(self):
        print("for strings")
class integer(base):
    def define(self):
        print("for int")
a=string()
a.define()
b=integer()
b.define()
b.defin1()
```

```python
# Q4. How can we achieve data abstraction?
# Ans : Data Abstraction can be achieved with the help of abc module which help us in adding the decorators to the methods in base class for abstracting the methods, this help us in defining he set of methods in child classes.
# eg :
from abc import ABC
class flora(ABC):
    @abc.abstractmethod
    def roots(self):
        pass
    @abc.abstractmethod
    def leaves(self):
        pass
class shrubs(flora):
    def roots(self):
        print("fibrous root")
    def leaves(self):
        print("tap leaves")
a=shrubs()
a.roots()
a.leaves()
```
