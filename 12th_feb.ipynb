```python
# Q1. What is exception in python ? write difference between exceptions and syntax errors.
# Ans : Exception is a special case of dealing with the errors where errors or unexpected situations occur and are resolved at run-time.
#     : Syntax error : the errors where the interpreter fails to understand the code and the code is synatically wrong and logically correct, this error occur at compile time.
#     : Exceptions : the error that is loicaly wrong and syntatically correct, this error occurs at run-time, these can be seperately handeled by try-except blocks.
# eg :
# print "hi" -> is a syntax error
# print(1/0) -> is an exception
```

```python
# Q2. What happens when exceptions are not handled. Explain with example.
# Ans : Exceptions are special case of errors that occur at run-time, they eventually get executed but they do not provide the required result and at times end up giving false results.
# eg : without using exceptions
n=int(input("enter a num for finding square :"))
print(n**2)
print("code executed")
# in the above case the code will be executed only if integer is entered as an input , and if integer is not entered then the last line which is independent of above two does not get executed due to the exception.
# eg : using exceptions
try :
    n=int(input("enter a num for finding square :"))
    print(n**2)
except :
    print("enter an int")
finally :
    print("code executed successfully")
# in the above code will not give any error irrespective of the input entered by the user.
```

```python
# Q3. Which python statments are used to catch and handle exceptions? Explain with examples.
# Ans : Try and except statements are used to catch and handle exceptions in Python. Statements that can raise exceptions are kept inside the try clause and the statements that handle the exception are written inside except clause.
try :
    n=int(input("enter a number to divide with"))
    a=100/n
except Exception as e :
    print(e)
```

```python
# Q4. Explain with an example :
# a : try and else
# b : finally
# c : raise
# Ans : 
# a : try and else :
# The else statement in Python is used in conjunction with the try and except statements to handle errors. The else statement is executed only if the try block does not raise an exception. This can be useful for executing code that should only be executed if there are no errors.
try :
    a=input("enter string 1:")
    b=input("enter string 2:")
    print(a+b)
except :
    print(Exception)
else :
    print("code is successfully executed")
# b : finally :
# finally block gets exected without any confition, i.e. even if an exception occurs or not
try :
    n=100/0
except :
    print(Exception)
finally :
    print("no syntax errors")
# c : raise :
# raise keyword is used to raise the errors in user defined exceptions
class ept(Exception):
    def __init__(self,msg):
        self.msg=msg
    def ept(n):
        if n==0 :
            raise ept("cannot be divided with zero")
        else:
            print("can be divide")
try :
    n=int(input("enter num :"))
    ept(n)
except :
    print(Exception)
```
