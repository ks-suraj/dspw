```python
# Q3. What errors are defined in the ArithmeticError class? Explain any two with an example.
# Ans : ArithmeticError class is class for errors that occur during arithmetic operations.
# types :
#       : 1 : ZeroDivisonError
#       : 2 : OverFLowError
#       : 3 : FloatingPointError
# 1 : ZeroDivisionError :
# This error handles the case where a number cannot be divide with zero.
try:
    result = 10 / 0
except ZeroDivisionError as e:
    print(f"Caught an error: {e}")
# 2 : OverFlowError :
# This error handles the case where there is involvment of very large numbers.
import math

try:
    result = math.exp(1000)  
except OverflowError as e:
    print(f"Caught an error: {e}")
```

```python
# Q4. Why LookupError class is used? Explain with an example KeyError and IndexError.
# Ans : The LookupError is a base class where the key or index that is being searched in a group or collection of elements is unavailable, the error that occurs at that time is LookupError.
#     : KeyError : KeyError is generally associated with keys of dictionaries, when the specified key of a dictionary does not exist.
try :
    dict = {'a':"app",'b':"bullet",'c':"chatgpt"}
    dict['e']
except KeyError as e :
    print(f"Caught a KeyError: {e}")
#     : IndexError : IndexError is associated with a group of elements that can be retrived through index and for a specified index, if the index does not exist then IndexError is popped.
try :
    l=[1,2,3]
    l[5]
except IndexError as e:
    print(f"Caught an IndexError: {e}")
```
