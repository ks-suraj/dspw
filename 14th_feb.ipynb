```python
# Q6. Explain deadlock and race conditions.
# Ans : Deadlock : Occurs when two or more threads try to lock the same variable at the same time, or when threads lock different variables and then try to lock each other's variables. This can create a circular dependency where each thread is waiting for the other to release the resource it needs, and the entire software system can come to a halt.
#     : Race Conditions : Occurs when two or more threads or processes try to access the same resource or variable at the same time. This can lead to data corruption or bugs. For example, if two people try to deposit money into the same bank account online at the same time, the final amount may be incorrect due to a race condition
```
