```python
# Q1. Create a function which will take a list as an argument and return the product of all the numbers
# after creating a flat list.
# Use the below-given list as an argument for your function.
def fun(a):
    l=[]
    for i in a :
        if type(i) is list:
            for j in i:
                l.append(j)
        else :
            l.append(i)
    return l

def mul(l):
    prod=1
    for i in l:
        if type(i) is int :
            prod*=i
        elif (type(i) is tuple) or (type(i) is set):
            for j in i :
                prod*=j
        elif type(i) is dict :
            for j in i :
                if type(j) is int :
                    prod*=j
                else :
                    continue
            for j in i :
                if type(i[j]) is int :
                    prod *= i[j]
                elif type(i[j]) is list or type(i[j]) is tuple:
                    for k in i[j] :
                        if type(k) is int :
                            prod*=k
                        else:
                            continue 
                else :
                    continue
        else :
            continue
        
    return prod
            
A=[1,2,3,4, [44,55,66, True], False, (34,56,78,89,34), {1,2,3,3,2,1}, {1:34, "key2": [55, 67, 78, 89], 4: (45,22, 61, 34)}, [56, 'data science'], 'Machine Learning']  
mul(fun(A))
```
