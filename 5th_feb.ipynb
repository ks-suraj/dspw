```python
# Q3. Explain why the __init__() function is used. Give a suitable example.
# Ans : __init__()  function is used as a constructor to pass the details to a class, through which class variables can be initialized.
# example :
class class1:
    def __init__(self,name):
        self.name=name
    def meth (self) :
        return self.name
# a = class1() # error will come as one parameter is required
b = class1("python")
b.meth()
```

```python
# Q4. Why self is used in OOPs?
# Ans : "self" is used to bind the variables and the methods with the class that are defined in that particular class.
# example :
class class1:
    def __init__(self,name):
        self.name=name
    def meth (self) :
        return self.name
    def meth2():
        return self.name
b = class1("python")
b.meth() # this will execute
# b.meth2() # this will not execute
# In the above progrm "name" variable is bind to the class, similarly "meth" method is bind together with class but "meth2" is not bind with class, hence it will not be executed.
```

```python
# Q5. What is inheritance? Give an example of each type of inheritance.
# Ans : Inheritance : In Object oriented programming, Inheritance is a method of inheriting the properties or behavior of a class to another class to ease the execution and reduce the complexities of a class.
#     : Types of Inhreitance :
#     : 1 - Single level Inheritabce : Inheriting from single class
class cls1:
    def square (self,num):
        print ("square is :",num**2)
        
class cls2(cls1):
    def cube (self,num):
        print("cube is:", num**3)

o = cls2()
o.square(2)
o.cube(2)

#     : 2 - Multiple Inheritance : Inheriting from multiple classes
class cls3():
    def root (self,num):
        print("root is:",num**0.5)
        
class cls4(cls1,cls3):
    def half(self,num):
        print("half is:",num/2)
    
p=cls4()
p.square(3)
p.half(3)
p.root(3)

#     3 - Multi-level Inheritance : Inheriting properties from a class that has been inherited from an another class
class cls5(cls4):
    def double (self,num):
        print ("double is:",num*2)
        
i=cls5()
i.square(4)
i.half(4)
i.root(4)
i.double(4)

#     4 - heirarichal Inheritance : Creating multiple classes from a single class
class cls6(cls4):
    def duplicate (self,num):
        print("duplicate is:",num,num)
        
class cls7(cls4):
    def perc(self,num):
        print("percentage is :",num%100)
        
j=cls6()
k=cls7()
j.square(5)
j.half(5)
j.root(5)
j.duplicate(5)
k.perc(5)

#     5 - Hybrid Inheritance : Creating classes from more than one format of inheritance type
class cls8(cls4,cls2):
    def binary(self,num):
        print ("binary is:",bin(num))
        
t=cls8()
t.binary(8)
```
