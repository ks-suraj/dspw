```python
# Q1. What is a database? Differentiate between SQL and NoSQL databases.
# Ans : A database is an organized collection of data that's stored, maintained, and accessed electronically. Databases can be used for data storage, analysis, and management.
#     : SQL : Structeres Query Language, Uses a relational model, storing data in tables with predefined schemas. SQL databases are good for structured data and multi-row transactions.
#           : softwares : MYSQL,MSSQL,etc
#     : NOSQL : NOt Only SQL, Uses a non-relational model, storing data in formats like documents or JSON without a fixed schema. NoSQL databases are good for unstructured, semi-structured, and dynamic data.     
#             : softwares : MongoDB,etc
```

```python
# Q2. What is DDL? Explain why CREATE, DROP, ALTER, and TRUNCATE are used with an example.
# Ans : DDL : DDL stande for data defionition language, set of sql commands used to create, alter, delete the structure but not the data.
# CREATE : SQL command used to create table
import mysql.connector
mydb=mysql.connector.connect(host="localhost",user="abc",password="password")
cur=mydb.cursor()
cur.execute("create table if not exists test1.table(cl1 varchar(20), cl2 int, cl3 float);")
cur.execute("insert into test1.table values ('aa',1,5.0);")
cur.execute("insert into test1.table values ('ab',2,10.0);")
cur.execute("insert into test1.table values ('ac',3,15.0);")
mydb.commit()
# ALTER : it is used to add, edit, deleter columns from the table.
cur.execute("create table if not exists test1.table2(cl1 varchar(20), cl2 int, cl3 float);")
cur.execute("alter table test1.table2 add cl5 varchar(10);")
mydb.commit()
# TRUNCATE : it is used to remove all the rows from the table i.e. erase data but structure stays intact.
cur.execute("truncate table test1.table2 ;")
# DROP : drop is used to remove the table and completly erase the structure or blueprint form the database.
cur.execute ("drop table test1.table2;")
```

```python
# Q3. What is DML? Explain INSERT, UPDATE, and DELETE with an example.
# Ans : DML stands for data manipulation language which is used retrive and manipulate data.
# INSERT : insert command is used to insert the vlaues in tha data.
import mysql.connector
mydb=mysql.connector.connect(host="localhost",user="abc",password="password")
cur=mydb.cursor()
cur.execute("create table if not exists test1.table3(cl1 varchar(20), cl2 int, cl3 float);")
cur.execute("insert into test1.table3 values ('aa',1,5.0);")
cur.execute("insert into test1.table3 values ('ab',2,10.0);")
cur.execute("insert into test1.table3 values ('ac',3,15.0);")
mydb.commit()
# UPDATE : update command is used to update and change the data.
cur.execute("update test1.table3 set cl1='aaa' where cl2=1 ;")
# DELETE : delete command is used to delete the data.
cur.execute ("delete from test1.table3 where cl2=1 ;")
mydb.commit()
```

```python
# Q4. What is DQL? Explain SELECT with an example.
# Ans : DQL stands for data query language which is used to processing and pergorming the queries.
# SELECT : select command is used to select the data and perform neccessary actions and display.
import mysql.connector
mydb=mysql.connector.connect(host="localhost",user="abc",password="password")
cur=mydb.cursor()
print(cur.execute("select * from test1.table3 ;"))

```

```python
# Q5. Explain primary key and foreign key.
# Ans : Primary key - A primary key is a unique identifier for a record in a database table, ensuring each entry is distinct.
#     : Foreign key - A **foreign key** is a column in a table that creates a relationship by referencing the primary key of another table.
```

```python
# Q7. Give the order of execution of SQL clauses in an SQL query.
# Ans : 
# the correct order of execution:
# FROM: Identifies the table(s) from which to retrieve data.
# JOIN: Combines rows from different tables based on a related column.
# WHERE: Filters rows based on specified conditions.
# GROUP BY: Groups rows that have the same values in specified columns.
# HAVING: Filters groups based on conditions (used with GROUP BY).
# SELECT: Specifies the columns to retrieve.
# DISTINCT: Removes duplicate rows from the result set.
# ORDER BY: Sorts the result set based on specified columns.
# LIMIT/OFFSET: Limits the number of rows returned and skips a specified number of rows
```
